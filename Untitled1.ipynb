```python
images = tf.data.Dataset.list_files('Face_detection_proj/train/images/*.jpg', shuffle=False)
```

```python
images.as_numpy_iterator().next()
```

```python
def load_image(x):
    byte_img = tf.io.read_file(x)
    img = tf.io.decode_jpeg(byte_img)
    return img
```

```python
images = images.map(load_image)
```
